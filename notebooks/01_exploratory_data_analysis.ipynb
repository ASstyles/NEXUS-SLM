# NEXUS-SLM: Exploratory Data Analysis & Leakage Audit

This notebook analyzes the multilingual support dataset, inspects class & language distributions,
and audits cross-split leakage between training and locked test sets.

```python
import json
import pandas as pd
from collections import Counter
from pathlib import Path

# Load datasets
train_df = pd.read_json('data/train/train.jsonl', lines=True)
val_df = pd.read_json('data/validation/validation.jsonl', lines=True)
test_df = pd.read_json('data/test/test.jsonl', lines=True)

print(f'Train size: {len(train_df)}, Val size: {len(val_df)}, Locked Test size: {len(test_df)}')
```

```python
# Category distribution
print('Category Distribution (Train):')
print(train_df['category'].value_counts())

print('\nLanguage Distribution (Train):')
print(train_df['language'].value_counts())
```

```python
# Inspect Leakage Report
with open('data/metadata/leakage_report.json', 'r') as f:
    leak_report = json.load(f)
print(json.dumps(leak_report, indent=2))
```
